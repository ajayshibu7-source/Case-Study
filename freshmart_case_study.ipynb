```python
q1=df['Quantity'].quantile(0.25)
q3=df['Quantity'].quantile(0.75)
iqr=q3-q1
lower_limit=q1-1.5*iqr
upper_limit=q3+1.5*iqr
outliers=df[(df['Quantity']<lower_limit)|(df['Quantity']>upper_limit)]
print("q1 ",q1)
print("q3 ",q3)
print("lowerlimit ",lower_limit)
print("upperlimint ",upper_limit)
print("outliers :\n",outliers)
```

```python
df_clean=df[~((df['Quantity']<lower_limit)|(df['Quantity']>upper_limit))].copy()
```

```python
df_clean["Gross"]=df_clean['Quantity']*df_clean['UnitPrice']
df_clean["Revenue"]=df_clean['Gross']*(1-df_clean['Discount']/100.0)
print(df_clean)
```

```python
df_clean['OrderSize']=df_clean['Revenue'].apply(lambda r: "Large" if r>=10000 else "Medium" if (r>=2000)&(r<10000) else "Small")
print(df_clean['OrderSize'].value_counts())
```
