```python
import pandas as pd

enrol_list = [
    "Python", "SQL", "Excel", "Python", "Power BI", "SQL", "Python",
    "Excel", "Tableau", "Python", "SQL", "Python", "Power BI", "Excel",
    "Python", "SQL", "Statistics", "Python", "Excel", "SQL", "Power BI",
    "Python", "SQL", "Excel", "Python", "Tableau", "SQL", "Power BI",
    "Python", "Excel", "Python", "SQL", "Power BI", "Statistics", "Python",
    "Tableau", "SQL", "Excel", "Machine Learning", "Power BI"
]

mode_list = [
    "Online", "Classroom", "Online", "Online", "Online", "Online", "Hybrid",
    "Online", "Online", "Online", "Classroom", "Online", "Hybrid", "Online",
    "Classroom", "Online", "Hybrid", "Hybrid", "Hybrid", "Hybrid", "Online",
    "Hybrid", "Online", "Hybrid", "Classroom", "Online", "Online", "Online",
    "Classroom", "Online", "Hybrid", "Classroom", "Hybrid", "Online", "Classroom",
    "Online", "Online", "Hybrid", "Online", "Online"
]

rating_list = [5, 4, 4, 3, 5, 2, 4, 5, 3, 4, 1, 5, 4, 3, 4, 5, 2, 4, 5, 4, 3, 4, 5, 2, 4]
```

```python
enrolments=pd.Series(enrol_list)
enrolments.name="Course Enrolled"
print(enrolments)
```

```python
print("total number of records =",enrolments.count())
```

```python
print(f"the first 5 records\n {enrolments.head(5)}")
print()
print(f"the last 5 records\n {enrolments.tail(5)}")
```

```python
c=enrolments.value_counts()
print(f"the numbers of enrolments received by each course \n {c}")
```

```python
share=round(enrolments.value_counts(normalize=True)*100,1)
print(f"each course's share of total enrolments as a percentage\n {share}")
```

```python
mode=pd.Series(mode_list)
mode.name="Study Mode"
print(mode)
```

```python
m=mode.value_counts()
shares=mode.value_counts(normalize=True)
print(f"the numbers of learners {m}\n percentage of share \n {shares*100}")
```

```python
print(f"the most preferred study mode = \n{m.head(1)} ")
print(f"combined percentage share of learns who study online  or hybrid = {shares.head(2).sum()*100}")
```

```python
rate=pd.Series(rating_list)
rate.name="ratings"
print(rate.name)
print(f'the number of ratings {rate.count()} ')
print(f' average ratings {round(rate.mean(),2)} ')
print(f'highest ratings {rate.max()} ')
print(f' lowest ratings {rate.min()} ')
```

```python
print(f'how many learners gave each rating score {rate.value_counts()} ')
```

```python
sentiment=lambda n: "Positive" if (n==4)|(n==5) else "Neutral" if (n==3) else "Negative" if (n==1)|(n==2) else "no rating"
s=rate.apply(sentiment)
print(f' category ratings \n {s} ')
```

```python
p=s.value_counts(normalize=True)*100
print(f'the percentage of learners in sentiment \n{round(p,1)} ')
```

```python
portfolio_action=lambda a: "Expand" if (a>=20) else "Maintain" if (a>=10)&(a<20) else "Review" if (a<10) else " "
pct=share.apply(portfolio_action)
print(f"{pct}")
```
