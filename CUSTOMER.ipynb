```python
import pandas as pd
```

```python
satisfaction_score=pd.Series([8,6,9,4,7,10,5,8])
print(satisfaction_score)
```

```python
satisfaction_score=pd.Series([8,6,9,4,7,10,5,8],index=["C001","C002","C003","C004","C005","C006","C007","C008"])
print(satisfaction_score)
```

```python
satisfaction_score=pd.Series({"C001":8,"C002":6,"C003":9,"C004":4,"C005":7,"C006":10,"C007":5,"C008":8})
print(satisfaction_score)
```

```python
print("values",satisfaction_score.values)
```

```python
satisfaction_score.name="Customer satisfaction"
print("name\n",satisfaction_score)
```

```python
print("score of customer c003 ",satisfaction_score.loc["C003"])
```

```python
print("score of customer 6th",satisfaction_score.iloc[5])
```

```python
print("score of customer C001 c003 C006 C008 \n ",satisfaction_score.loc[["C001","C003","C006","C008"]])
```

```python
print("score of customer C001 c003 C006 C008 \n ",satisfaction_score.iloc[[0,2,5,7]])
```

```python
print("score of customer  C002 AND C007 \n ",satisfaction_score.loc[["C002","C007"]])
```

```python
print("total numbers of  customers =",satisfaction_score.count())
```

```python
print("total satisfaction score=",satisfaction_score.sum())
```

```python
print("average satisfaction score=",satisfaction_score.mean())
```

```python
print("highest satisfaction score=",satisfaction_score.max())
```

```python
print("lowest satisfaction score=",satisfaction_score.min())
```

```python
print("median satisfaction score=",satisfaction_score.median())
```

```python
print("all customers whose satisfaction score is greater than or equal 8 =\n",satisfaction_score[satisfaction_score>=8])
```

```python
print("all customers whose satisfaction score is less than 6 =\n",satisfaction_score[satisfaction_score<6])
```
