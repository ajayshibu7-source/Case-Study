```python
print(f'total sales : {df['Sales'].sum()}')
print(f'total profit : {df['Profit'].sum()}')
print(f'average order-line sales  : {df['Sales'].mean()}')
```

```python
category_sales=df.groupby("Category")['Sales'].sum()
print(f'categories generate the most sales :{category_sales.nlargest(1)}')
```

```python
sub_cat=df.groupby('Sub-Category')['Profit'].sum().sort_values(ascending=False)
print(f'highest profit category{sub_cat.head(1)}')
print(f'lowest profit category{sub_cat.tail(1)}')
```

```python
region_sales=df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
print(f'highest sales region {region_sales.head(1)}')
```

```python
segment_sales=df.groupby('Segment')['Sales'].sum().sort_values(ascending=False)
print(f'most sales customer segment {segment_sales.head(1)}')
```

```python

print(f'discount related to profit: {df['Discount'].corr(df['Profit'])}')
```

```python
discount_impact = df.groupby('Discount')['Profit'].mean()
print("\nAverage Profit by Discount Level:\n", discount_impact)
```

```python

df['Order Date'] = pd.to_datetime(df['Order Date'])
monthly_sales = df.groupby(df['Order Date'].dt.to_period('M'))['Sales'].sum()
print("Monthly Sales Trend:\n", monthly_sales)
```
