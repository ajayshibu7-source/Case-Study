```python
import pandas as pd

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
revenue_values  = [42, 45, 50, 48, 55, 60, 58, 62, 65, 70, 75, 85]
expense_values  = [38, 40, 42, 47, 50, 52, 55, 60, 61, 64, 66, 72]
```

```python
revenue=pd.Series(revenue_values,index=months,name="revenue")
expense=pd.Series(expense_values,index=months,name="expense")
print(revenue)
print(expense)
```

```python
#If a = [42, 45, 50] and b = [38, 40, 42], evaluating a + b results in list concatenation ([42, 45, 50, 38, 40, 42]) because Python lists treat + as sequence merging. Conversely, Pandas Series perform element-wise vectorised arithmetic aligned automatically by index labels.
a = [42, 45, 50]
b = [38, 40, 42]
print(a+b)
```

```python
profit_list=[]
for r,e in zip(revenue_values,expense_values):
  profit_list.append(r-e)
print(profit_list)
```

```python
profit=revenue-expense
profit.name="profit"
print(f"monthly profit \n {profit}")
```

```python
margin=round(profit/revenue*100,1)
margin.name="profit margin %"
print(f"margin\n {margin}")
```

```python
total_revenue=revenue.sum()
total_expenses=expense.sum()
total_profit=profit.sum()
overall_margin=round((total_profit/total_revenue)*100,1)
print(f"total revenue = {total_revenue}")
print(f"total profit = {total_profit}")
print(f"total expenses = {total_expenses}")
print(f"overall_margin = {overall_margin}")
```

```python
q1=profit.loc['Jan':'Mar'].sum()
q2=profit.loc['Apr':'Jun'].sum()
q3=profit.loc['Jul':'Sep'].sum()
q4=profit.loc['Oct':'Dec'].sum()
q=pd.Series([q1,q2,q3,q4])
print(f"Q1 Profit: {q1} | Q2 Profit: {q2} | Q3 Profit: {q3} | Q4 Profit: {q4}")
print(f"Strongest Quarter: (Profit: {q.max()})")
```

```python
h1=profit.iloc[0:6].sum()
h2=profit.iloc[6:12].sum()
print(f"h1 profit {h1} \n h2 profit {h2}")
```

```python
festival_profit=profit.loc[['Oct','Nov','Dec']].sum()
print(f"\nFestive Months Profit (Oct-Dec):{festival_profit}")
```

```python
# iloc equivalent to loc['Apr':'Jun']
# loc['Apr':'Jun'] includes indices 'Apr', 'May', 'Jun' (positions 3, 4, 5).
# Equivalent iloc expression: profit.iloc[3:6] (stop index 6 is exclusive).
print(profit.loc['Apr':'Jun'])
print(profit.iloc[3:5])
```

```python
print(f"months where profit margin exceeds 10%.{margin[margin>10.0]}")
```

```python
print(f"months where profit is below 4 lakh \n{profit[profit<4]} ")
```

```python
ranked_profit=profit.sort_values(ascending=False)

print(f"top 3 months by profit\n {ranked_profit[:3]}")
```

```python
avg_profit=profit.mean()
print(f'the average monthly profit {avg_profit}')
print(f'the average monthly profit that exceeds\n {profit[profit>avg_profit]}')
```

```python
p=lambda n: "Strong" if (n>=8) else "Stable" if(n<8)&(n>=4) else "Weak" if (n<4) else "fail"
profit_status=profit.apply(p)
print(f'profit_status \n {profit_status}')
```

```python
status_counts=profit_status.value_counts()
print(f"number of status {status_counts}")
status_pct=profit_status.value_counts(normalize=True)*100
print(f"percentage of months in each status\n{status_pct}")
```

```python
q=lambda n:0.25*n if (n>0) else "0"
tax_payable=profit.apply(q)
print(f"tax  \n{tax_payable}")

```

```python
total_tax=tax_payable.sum()
print(f"total tax {total_tax}")
post_tax=total_profit-total_tax
print(f"post tax profit {post_tax}")
```
