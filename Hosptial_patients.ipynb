```python
q1=df1['BillAmount'].quantile(0.25)
q3=df1['BillAmount'].quantile(0.75)
iqr=q3-q1
lower_limit=q1-1.5*iqr
upper_limit=q3+1.5*iqr
outliers=df1[(df1['BillAmount']<lower_limit)|(df1['BillAmount']>upper_limit)]
print("q1 ",q1)
print("q3 ",q3)
print("iqr ",iqr)
print("lowerlimit ",lower_limit)
print("upperlimit ",upper_limit)
print("outliers \n",outliers)
```

```python
df_clean=df1[~((df1['BillAmount']<lower_limit)|(df1['BillAmount']>upper_limit))].copy()
```

```python
df_clean.shape
```

```python
print("the correlation between StayDays and BillAmount\n",df_clean['StayDays'].corr(df_clean['BillAmount']))
```

```python
df_clean['Cost_Per_Day']=df_clean['BillAmount']/df_clean['StayDays']
```

```python
print("department has highest average cost per day\n",df_clean.groupby('Department')['Cost_Per_Day'].mean().idxmax())
```

```python
df_clean['Stay_Type']=df_clean['StayDays'].apply(lambda s:"Short" if s<=3 else "Medium" if (s>=4)&(s<=7) else "long" if s>7 else "no days")
print(df_clean)
```

```python
df_clean['Age_Group']=df_clean['Age'].apply(lambda a:"Child" if a<18 else "Adult" if (a>=18)&(a<=59) else "Senior" if a>=60 else "nothing")
print(df_clean)
```

```python
print("average SatisfactionScore for each Stay_Type\n",df_clean.groupby('Stay_Type')['SatisfactionScore'].mean())
```

```python
print("average SatisfactionScore for each Age_Group\n",df_clean.groupby('Age_Group')['SatisfactionScore'].mean())
```
