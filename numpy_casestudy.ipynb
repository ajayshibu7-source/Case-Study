```python
import numpy as np
stores=np.array(["store_A","store_B","store_c","store_d","store_e"])
sales=np.array([ [120000,135000,142000,155000],[98000,105000,110000,108000],[150000,148000,160000,172000],[87000,92000,99000,115000],[130000,138000,141000,149000]])
print(f'totalsales for every store : {sales.sum(axis=1)}')
print(f'averagequarterlysales for every store : {sales.mean(axis=1)}')
print(f'totalsales for every quarter : {sales.sum(axis=0)}')
totalsales=sales.sum(axis=1)
i=np.argmax(totalsales)
print(f'highest annualsales : {stores[i]}')
j=np.argsort(totalsales)[::-1]
print(f'rank highest to lowest : {stores[j]}')
overallaverage=np.mean(totalsales)
print(overallaverage)
flaggedstores=stores[totalsales>overallaverage]
print(f'flag stores ={flaggedstores}')


```

```python
students=np.array(["anu","binu","cathy","deepa","ebin","fahad"])
marks=np.array([[82,75,91,88],[61,68,70,65],[45,52,48,55],[90,94,89,92],[72,80,76,85],[38,42,49,44]])
subjects=np.array(["python","sql","excel","statistics"])
```

```python
totalmarks=np.sum(marks,axis=1)
print(f'totalmarks : {totalmarks}')
saveragemarks=np.mean(marks,axis=1)
print(f'averagemarks : {saveragemarks}')
print(f'{students} :{totalmarks}')
```

```python
averagemarks=np.mean(marks,axis=0)
print(f'averagemarks for subject : {averagemarks}')
```

```python
i=np.max(marks,axis=0)
print(i)
```

```python
i=np.where(marks>=50,"pass","fail")
print(f'{students} : marks : {i}')
```

```python
below_60_students=students[saveragemarks<60]
print(below_60_students)
```
