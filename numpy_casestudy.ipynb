```python
#retail sales performance
import numpy as np
stores=np.array(["store_A","store_B","store_c","store_d","store_e"])
sales=np.array([ [120000,135000,142000,155000],[98000,105000,110000,108000],[150000,148000,160000,172000],[87000,92000,99000,115000],[130000,138000,141000,149000]])
print(f'totalsales for every store : {sales.sum(axis=1)}')
print(f'averagequarterlysales for every store : {sales.mean(axis=1)}')
print(f'totalsales for every quarter : {sales.sum(axis=0)}')
totalsales=sales.sum(axis=1)
i=np.argmax(totalsales)
print(f'highest annualsales : {stores[i]}')
j=np.argsort(totalsales)[::-1]
print(f'rank highest to lowest : {stores[j]}')
overallaverage=np.mean(totalsales)
print(overallaverage)
flaggedstores=stores[totalsales>overallaverage]
print(f'flag stores ={flaggedstores}')
change=(np.diff(sales,axis=1)/sales[:,:-1])*100
print(f'quarter overquarter change = {np.round(change,2)}')

```

```python
# student performance analytics
```

```python
students=np.array(["anu","binu","cathy","deepa","ebin","fahad"])
marks=np.array([[82,75,91,88],[61,68,70,65],[45,52,48,55],[90,94,89,92],[72,80,76,85],[38,42,49,44]])
subjects=np.array(["python","sql","excel","statistics"])
```

```python
totalmarks=np.sum(marks,axis=1)
print(f'totalmarks : {totalmarks}')
saveragemarks=np.mean(marks,axis=1)
print(f'averagemarks : {saveragemarks}')
print(f'{students} :{totalmarks}')
```

```python
averagemarks=np.mean(marks,axis=0)
print(f'averagemarks for subject : {averagemarks}')
```

```python
i=np.max(marks,axis=0)
print(i)
```

```python
i=np.where(marks>=50,"pass","fail")
print(f'{students} : marks : {i}')
```

```python
below_60_students=students[saveragemarks<60]
print(below_60_students)
```

```python
conditions=[marks>=80,marks>=60,marks>=50]
choice=["A","B","C"]
grade=np.select(conditions,choice,default="F")
print(grade)
```

```python
emp = ["Arun", "Bhavana", "Charan", "Divya", "Eshan", "Fathima"]
perfor = np.array([
    [78, 82, 85, 88],
    [65, 70, 68, 74],
    [92, 89, 95, 91],
    [55, 62, 60, 67],
    [81, 79, 84, 87],
    [72, 75, 78, 80]
])
quarters = ["Q1", "Q2", "Q3", "Q4"]
print(emp)
```

```python
#total performance
total_perfor=np.sum(perfor,axis=1)
print(total_perfor)
```

```python
#average quaterly performace for every employee
average=np.mean(perfor,axis=1)
print(f'Average : ',average)
```

```python
#averageperformance for each quarter
qaverage=np.mean(perfor,axis=0)
print(f'Average quarter: ',qaverage)
```

```python
#employee with highest total performance
h=np.argmax(total_perfor)
print(f'highest total performance is {emp[h]}')
```

```python
#rank all employee
r=np.argsort(total_perfor[::-1])
for i in r:
 print(emp[i],":",total_perfor[i])
```

```python
#percentage growth
growth=np.diff(perfor,axis=1)/perfor[:,:-1]*100
print("Growth:",np.round(growth, 2))
```

```python
#avg equal or less than 80
print("avg equal or less than 80 :",np.array(emp)[average>=80])
```

```python
print(f'no of employee have average below 70 : {np.size(average[average<70])}')
```

```python
#highest performance
print(np.max(perfor))
```

```python
#employee and quater where highest score
row,col= np.unravel_index(np.argmax(perfor), perfor.shape)

print("Employee:", emp[row])
print("Quarter:", quarters[col])
```
